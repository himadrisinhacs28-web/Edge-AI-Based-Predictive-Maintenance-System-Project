```python
import os
import random

data_dir = "/kaggle/input/mafaulda-full/mafaulda"

classes = {
    "normal": 0,
    "imbalance": 1,
    "horizontal-misalignment": 2,
    "vertical-misalignment": 3
}

random.seed(42)

train_files_new = []
val_files = []
test_files = []

for class_name, label in classes.items():
    folder = os.path.join(data_dir, class_name)

    # Find CSV files in all nested folders
    files = []

    for root, dirs, filenames in os.walk(folder):
        for filename in filenames:
            if filename.lower().endswith(".csv"):
                files.append(os.path.join(root, filename))

    files.sort()
    random.shuffle(files)

    n = len(files)
    n_train = int(n * 0.64)
    n_val = int(n * 0.16)

    train_part = files[:n_train]
    val_part = files[n_train:n_train + n_val]
    test_part = files[n_train + n_val:]

    train_files_new.extend((f, label) for f in train_part)
    val_files.extend((f, label) for f in val_part)
    test_files.extend((f, label) for f in test_part)

    print(
        f"{class_name}: total={n}, "
        f"train={len(train_part)}, "
        f"validation={len(val_part)}, "
        f"test={len(test_part)}"
    )

print("\nTotal files:")
print("Train:", len(train_files_new))
print("Validation:", len(val_files))
print("Test:", len(test_files))

# Verify no file appears in multiple splits
train_paths = {f for f, label in train_files_new}
val_paths = {f for f, label in val_files}
test_paths = {f for f, label in test_files}

print("\nTrain-Val overlap:", len(train_paths & val_paths))
print("Train-Test overlap:", len(train_paths & test_paths))
print("Val-Test overlap:", len(val_paths & test_paths))

# Stop if any class is missing
assert all(
    sum(label == class_id for _, label in split) > 0
    for split in [train_files_new, val_files, test_files]
    for class_id in classes.values()
), "ERROR: A class is missing from one of the splits!"

print("\nSUCCESS: All selected classes are present in all splits.")
```

```python
import numpy as np
import pandas as pd
from scipy.stats import skew, kurtosis

def extract_time_features(signal):
    rms = np.sqrt(np.mean(signal ** 2))
    peak = np.max(np.abs(signal))

    return [
        np.mean(signal),
        np.std(signal),
        rms,
        np.mean(np.abs(signal)),
        np.ptp(signal),
        peak,
        peak / (rms + 1e-8),
        skew(signal),
        kurtosis(signal)
    ]

feature_rows = []

for window in vibration_windows:
    row = []

    # Three vibration axes
    for axis in range(3):
        row.extend(extract_time_features(window[:, axis]))

    feature_rows.append(row)

feature_names = [
    f"Axis_{axis}_{name}"
    for axis in [4, 5, 6]
    for name in [
        "Mean", "Std", "RMS", "Mean_Absolute",
        "Peak_to_Peak", "Peak", "Crest_Factor",
        "Skewness", "Kurtosis"
    ]
]

X_sample = pd.DataFrame(feature_rows, columns=feature_names)

print("Extracted feature shape:", X_sample.shape)
display(X_sample.head())
```

```python
FS = 50000  # Sampling frequency: 50 kHz

def extract_fft_features(signal, fs=FS):
    signal = signal - np.mean(signal)

    spectrum = np.abs(np.fft.rfft(signal))
    frequencies = np.fft.rfftfreq(
        len(signal), d=1 / fs
    )

    # DC component remove karo
    spectrum[0] = 0

    power = spectrum ** 2
    total_power = np.sum(power) + 1e-12

    # Five frequency bands
    edges = np.linspace(0, fs / 2, 6)
    band_energy = []

    for i in range(5):
        mask = (
            (frequencies >= edges[i]) &
            (frequencies < edges[i + 1])
        )
        band_energy.append(
            np.sum(power[mask]) / total_power
        )

    dominant_freq = frequencies[np.argmax(spectrum)]
    spectral_centroid = (
        np.sum(frequencies * spectrum) /
        (np.sum(spectrum) + 1e-12)
    )

    return [
        dominant_freq,
        spectral_centroid,
        np.sqrt(np.mean(signal ** 2)),
        np.max(spectrum),
        np.sum(power),
        *band_energy
    ]

# Sample file ke har window aur axis par FFT
fft_rows = []

for window in vibration_windows:
    row = []

    for axis in range(3):
        row.extend(
            extract_fft_features(window[:, axis])
        )

    fft_rows.append(row)

fft_names = [
    f"Axis_{axis}_{name}"
    for axis in [4, 5, 6]
    for name in [
        "Dominant_Frequency",
        "Spectral_Centroid",
        "FFT_RMS",
        "Max_Spectrum",
        "Total_Power",
        "Band_1_Energy",
        "Band_2_Energy",
        "Band_3_Energy",
        "Band_4_Energy",
        "Band_5_Energy"
    ]
]

X_fft_sample = pd.DataFrame(
    fft_rows, columns=fft_names
)

print("FFT feature shape:", X_fft_sample.shape)
display(X_fft_sample.head())
```

```python
# Time-domain aur FFT features combine karo
X_combined_sample = pd.concat(
    [X_sample, X_fft_sample],
    axis=1
)

print("Combined feature shape:", X_combined_sample.shape)
print("Total features:", len(X_combined_sample.columns))
print("Missing values:", X_combined_sample.isnull().sum().sum())
print(
    "Infinite values:",
    np.isinf(X_combined_sample.to_numpy()).sum()
)

display(X_combined_sample.head())
```

```python

import os
import numpy as np
import pandas as pd
from scipy.stats import skew, kurtosis

WINDOW_SIZE = 1024
FS = 50000
AXIS_COLUMNS = [4, 5, 6]

def extract_features_from_file(file_path, label):
    df_file = pd.read_csv(file_path, header=None)

    signals = df_file[AXIS_COLUMNS].to_numpy(dtype=np.float64)
    n_windows = len(signals) // WINDOW_SIZE

    if n_windows == 0:
        return [], []

    signals = signals[:n_windows * WINDOW_SIZE]
    windows = signals.reshape(n_windows, WINDOW_SIZE, 3)

    rows = []

    for window in windows:
        row = []

        for axis in range(3):
            x = window[:, axis]

            # Time-domain features
            rms = np.sqrt(np.mean(x ** 2))
            peak = np.max(np.abs(x))

            row.extend([
                np.mean(x),
                np.std(x),
                rms,
                np.mean(np.abs(x)),
                np.ptp(x),
                peak,
                peak / (rms + 1e-8),
                skew(x),
                kurtosis(x)
            ])

            # FFT features
            centered = x - np.mean(x)
            spectrum = np.abs(np.fft.rfft(centered))
            frequencies = np.fft.rfftfreq(
                len(centered), d=1 / FS
            )

            spectrum[0] = 0
            power = spectrum ** 2
            total_power = np.sum(power) + 1e-12

            edges = np.linspace(0, FS / 2, 6)
            band_energy = []

            for i in range(5):
                mask = (
                    (frequencies >= edges[i]) &
                    (frequencies < edges[i + 1])
                )
                band_energy.append(
                    np.sum(power[mask]) / total_power
                )

            dominant_freq = frequencies[np.argmax(spectrum)]
            centroid = (
                np.sum(frequencies * spectrum) /
                (np.sum(spectrum) + 1e-12)
            )

            row.extend([
                dominant_freq,
                centroid,
                np.sqrt(np.mean(centered ** 2)),
                np.max(spectrum),
                np.sum(power),
                *band_energy
            ])

        rows.append(row)

    labels = [label] * len(rows)
    return rows, labels


print("Feature extraction function ready!")

```

```python

def process_file_list(file_list):
    all_features = []
    all_labels = []

    for i, (file_path, label) in enumerate(file_list):
        features, labels = extract_features_from_file(
            file_path, label
        )

        all_features.extend(features)
        all_labels.extend(labels)

        if (i + 1) % 50 == 0 or i + 1 == len(file_list):
            print(f"Processed {i + 1}/{len(file_list)} files")

    X = np.asarray(all_features, dtype=np.float32)
    y = np.asarray(all_labels, dtype=np.int64)

    return X, y


print("Processing training files...")

X_train, y_train = process_file_list(train_files_new)

print("\nTraining features shape:", X_train.shape)
print("Training labels shape:", y_train.shape)
```

```python

import numpy as np
import pandas as pd

WINDOW_SIZE = 1024

# Vibration column 5 aur acoustic column 7
VIBRATION_COL = 5
ACOUSTIC_COL = 7

def create_windows(file_path, label):
    df = pd.read_csv(file_path, header=None)

    vibration = df.iloc[:, VIBRATION_COL].to_numpy(
        dtype=np.float32
    )
    acoustic = df.iloc[:, ACOUSTIC_COL].to_numpy(
        dtype=np.float32
    )

    number_of_windows = len(df) // WINDOW_SIZE

    vibration_windows = []
    acoustic_windows = []
    labels = []

    for i in range(number_of_windows):
        start = i * WINDOW_SIZE
        end = start + WINDOW_SIZE

        vibration_windows.append(vibration[start:end])
        acoustic_windows.append(acoustic[start:end])
        labels.append(label)

    return (
        np.array(vibration_windows, dtype=np.float32),
        np.array(acoustic_windows, dtype=np.float32),
        np.array(labels, dtype=np.int32)
    )

# Ek training file par test karo
sample_path, sample_label = train_files[0]

X_vib_sample, X_aud_sample, y_sample = create_windows(
    sample_path, sample_label
)

print("Sample file:", sample_path)
print("Label:", sample_label)
print("Vibration windows shape:", X_vib_sample.shape)
print("Acoustic windows shape:", X_aud_sample.shape)
print("Labels shape:", y_sample.shape)
print("Windows created:", len(y_sample))
```

```python

from sklearn.preprocessing import StandardScaler
import numpy as np

# Vibration aur acoustic data ki copies
X_vib_train_scaled = X_vib_train.copy()
X_vib_test_scaled = X_vib_test.copy()

X_aud_train_scaled = X_aud_train.copy()
X_aud_test_scaled = X_aud_test.copy()

# Har window ko separately standardize karo
def normalize_windows(train_data, test_data):
    train_mean = train_data.mean(axis=1, keepdims=True)
    train_std = train_data.std(axis=1, keepdims=True)
    train_std[train_std < 1e-8] = 1.0

    test_mean = test_data.mean(axis=1, keepdims=True)
    test_std = test_data.std(axis=1, keepdims=True)
    test_std[test_std < 1e-8] = 1.0

    train_scaled = (train_data - train_mean) / train_std
    test_scaled = (test_data - test_mean) / test_std

    return train_scaled.astype(np.float32), test_scaled.astype(np.float32)

X_vib_train_scaled, X_vib_test_scaled = normalize_windows(
    X_vib_train, X_vib_test
)

X_aud_train_scaled, X_aud_test_scaled = normalize_windows(
    X_aud_train, X_aud_test
)

print("Vibration training shape:", X_vib_train_scaled.shape)
print("Acoustic training shape:", X_aud_train_scaled.shape)
print("Vibration mean of first window:", round(float(X_vib_train_scaled[0].mean()), 4))
print("Vibration std of first window:", round(float(X_vib_train_scaled[0].std()), 4))
```

```python

def build_cnn_model(input_shape=(1024, 1), num_classes=4):
    model = keras.Sequential([
        keras.Input(shape=input_shape),

        layers.Conv1D(16, kernel_size=7, activation="relu"),
        layers.MaxPooling1D(pool_size=2),

        layers.Conv1D(32, kernel_size=5, activation="relu"),
        layers.MaxPooling1D(pool_size=2),

        layers.Conv1D(64, kernel_size=3, activation="relu"),
        layers.GlobalAveragePooling1D(),

        layers.Dense(32, activation="relu"),
        layers.Dropout(0.3),

        layers.Dense(num_classes, activation="softmax")
    ])

    model.compile(
        optimizer="adam",
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model


vibration_model = build_cnn_model()

vibration_model.summary()
```

```python

import numpy as np
import pandas as pd

from scipy.stats import skew, kurtosis
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report


def extract_features_from_windows(windows):
    features = []

    for x in windows:
        rms = np.sqrt(np.mean(x ** 2))
        std = np.std(x)
        peak = np.max(np.abs(x))
        mean_abs = np.mean(np.abs(x))
        peak_to_peak = np.ptp(x)
        crest_factor = peak / (rms + 1e-8)
        zero_crossings = np.mean(np.diff(np.signbit(x)))

        features.append([
            np.mean(x),
            std,
            rms,
            mean_abs,
            peak,
            peak_to_peak,
            crest_factor,
            skew(x),
            kurtosis(x),
            zero_crossings
        ])

    return np.asarray(features, dtype=np.float32)


def build_feature_dataset(file_list, sensor_col):
    all_features = []
    all_labels = []

    for file_path, label in file_list:
        df = pd.read_csv(file_path, header=None)
        signal = df.iloc[:, sensor_col].to_numpy(dtype=np.float32)

        n_windows = len(signal) // WINDOW_SIZE

        for i in range(n_windows):
            start = i * WINDOW_SIZE
            window = signal[start:start + WINDOW_SIZE]

            feats = extract_features_from_windows(
                window.reshape(1, -1)
            )[0]

            all_features.append(feats)
            all_labels.append(label)

    return (
        np.asarray(all_features, dtype=np.float32),
        np.asarray(all_labels, dtype=np.int32)
    )


# Vibration signal features
Xf_train, yf_train = build_feature_dataset(
    train_files_new, VIBRATION_COL
)

Xf_val, yf_val = build_feature_dataset(
    val_files, VIBRATION_COL
)

Xf_test, yf_test = build_feature_dataset(
    test_files, VIBRATION_COL
)

print("Training features:", Xf_train.shape)
print("Validation features:", Xf_val.shape)
print("Testing features:", Xf_test.shape)
```

```python

import numpy as np
import pandas as pd

from scipy.stats import skew, kurtosis
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

FS = 50000  # MAFAULDA sampling frequency: 50 kHz

def extract_fft_features(x, fs=FS):
    x = x - np.mean(x)

    # One-sided frequency spectrum
    spectrum = np.abs(np.fft.rfft(x))
    freqs = np.fft.rfftfreq(len(x), d=1/fs)

    # Ignore DC component
    spectrum[0] = 0

    power = spectrum ** 2
    total_power = np.sum(power) + 1e-12

    # Five broad frequency-band energy ratios
    edges = np.linspace(0, fs / 2, 6)
    band_energy = []

    for i in range(5):
        mask = (freqs >= edges[i]) & (freqs < edges[i+1])
        band_energy.append(
            np.sum(power[mask]) / total_power
        )

    dominant_freq = freqs[np.argmax(spectrum)]
    centroid = np.sum(freqs * spectrum) / (
        np.sum(spectrum) + 1e-12
    )

    return [
        dominant_freq,
        centroid,
        np.sqrt(np.mean(x**2)),
        np.max(spectrum),
        np.sum(power),
        *band_energy
    ]


def extract_45_features(window):
    features = []

    # Three overhang vibration axes: columns 4, 5, 6
    for col in range(3):
        x = window[:, col]

        rms = np.sqrt(np.mean(x**2))
        peak = np.max(np.abs(x))

        # 9 time-domain features
        features.extend([
            np.mean(x),
            np.std(x),
            rms,
            np.mean(np.abs(x)),
            np.ptp(x),
            peak,
            peak / (rms + 1e-8),
            skew(x),
            kurtosis(x)
        ])

        # 8 frequency-domain features per axis
        features.extend(extract_fft_features(x))

    return np.nan_to_num(
        np.asarray(features, dtype=np.float32),
        nan=0.0, posinf=0.0, neginf=0.0
    )


def build_fft_dataset(file_list):
    X, y = [], []

    for file_path, label in file_list:
        df = pd.read_csv(file_path, header=None)
        signal = df.iloc[:, [4, 5, 6]].to_numpy(
            dtype=np.float32
        )

        n_windows = len(signal) // WINDOW_SIZE

        for i in range(n_windows):
            start = i * WINDOW_SIZE
            window = signal[start:start + WINDOW_SIZE]

            X.append(extract_45_features(window))
            y.append(label)

    return np.asarray(X, dtype=np.float32), np.asarray(y)


# Build datasets using existing file-wise splits
Xfft_train, yfft_train = build_fft_dataset(train_files_new)
Xfft_val, yfft_val = build_fft_dataset(val_files)
Xfft_test, yfft_test = build_fft_dataset(test_files)

print("Train:", Xfft_train.shape)
print("Validation:", Xfft_val.shape)
print("Test:", Xfft_test.shape)

# Train model
rf_fft = RandomForestClassifier(
    n_estimators=200,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

rf_fft.fit(Xfft_train, yfft_train)

val_pred_fft = rf_fft.predict(Xfft_val)
test_pred_fft = rf_fft.predict(Xfft_test)

print("\nValidation accuracy:",
      round(accuracy_score(yfft_val, val_pred_fft) * 100, 2), "%")

print("Test accuracy:",
      round(accuracy_score(yfft_test, test_pred_fft) * 100, 2), "%")

print("\nTest classification report:")
print(classification_report(
    yfft_test,
    test_pred_fft,
    labels=[0, 1, 2, 3],
    target_names=[
        "Normal", "Imbalance",
        "Horizontal Misalignment", "Vertical Misalignment"
    ],
    zero_division=0
))
```
